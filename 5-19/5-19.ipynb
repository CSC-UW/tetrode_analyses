## Reflect the channel map and tetrode grouping on the SpikeInterface extractor

The raw `extractor` above is a flat list of 64 channels (`CH1`…`CH64`) with no probe
and no grouping. `ta.attach_tetrode_probegroup` reorders the channels into channel-map
order and attaches a probeinterface `ProbeGroup` (one tetrode per group), giving the
recording a `group` property. The recording can then be split / preprocessed / sorted
per tetrode (see SpikeInterface's *Work with tetrodes* and *Process a recording by
channel group* how-tos), and its tetrode structure matches the `ta` DataArray above.

```python
grouped = ta.attach_tetrode_probegroup(extractor)
grouped
```

```python
# 0-based tetrode index per channel: [0, 0, 0, 0, 1, 1, 1, 1, ...]
grouped.get_channel_groups()
```

```python
by_tt = grouped.split_by("group")
print(f"{len(by_tt)} per-tetrode recordings")
by_tt[0].get_channel_ids()  # TT1 = CH40, CH38, CH36, CH34 (channel-map order)
```

```python
# The extractor grouping now matches the `ta` DataArray's tetrode structure.
da_flat = ta.open_tetrode_dataarray(ACQ_DIR, layout="2d_flat_index")
tt1_from_da = list(da_flat.channel.values[:4])
tt1_from_si = list(map(str, by_tt[0].get_channel_ids()))
print("ta TT1:", tt1_from_da)
print("si TT1:", tt1_from_si)
assert tt1_from_si == tt1_from_da
print("tetrode property:", list(map(str, grouped.get_property("tetrode")[:8])))
```
